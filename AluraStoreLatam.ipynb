### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

```python
import matplotlib.pyplot as plt
```

#1. Análisis de facturación



```python
precioTotal1= float(tienda['Precio'].sum())
precioTotal2= float(tienda2['Precio'].sum())
precioTotal3= float(tienda3['Precio'].sum())
precioTotal4= float(tienda4['Precio'].sum())

print(precioTotal1, precioTotal2, precioTotal3, precioTotal4)

tiendas=["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]
preciosTotales=[precioTotal1, precioTotal2, precioTotal3, precioTotal4]

mayorIngreso=max(preciosTotales)
menorIngreso=min(preciosTotales)
print(f"La tienda que mejor vende se encuentra en la posicion: ", preciosTotales.index(mayorIngreso), "y la que menos vende se encuentra en la posicion:", preciosTotales.index(menorIngreso))

plt.bar(tiendas, preciosTotales, color= ['skyblue', 'pink', 'black', 'green'])
plt.title("Ventas totales en cada tienda")
plt.xlabel("Tiendas")
plt.ylabel("Total de ventas")
plt.show()
```

# 2. Ventas por categoría

```python
#Separacion de categorias y cuántas veces aparece (ventas) en tienda 1
def filtradoCantidadCategorias(tiendaJuan):
  cantidadCategorias={}
  for categoria in tiendaJuan:
    if categoria in cantidadCategorias:
      cantidadCategorias[categoria] +=1
    else:
      cantidadCategorias[categoria]= 1

  for categoria, numero in cantidadCategorias.items():
    print(f"{categoria} aparece {numero}")

#Categoria más vendida
  categoriaMasVendida=max(cantidadCategorias, key=cantidadCategorias.get)
  print(f"La categoria más vendida en Tienda 1 es:", categoriaMasVendida)
  return cantidadCategorias

ventasPorCategoria1=filtradoCantidadCategorias(tienda['Categoría del Producto'])

categorias=list(ventasPorCategoria1.keys())
ventas=list(ventasPorCategoria1.values())

plt.pie(ventas, labels=categorias, autopct='%1.1f%%')
plt.title("Ventas por categoria en tienda 1")
plt.show()
```
